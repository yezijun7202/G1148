Actualment, Python treballa amb **Unicode**, que permet representar caràcters
de pràcticament tots els sistemes d'escriptura.

Una cadena `str` de Python és una seqüència de **punts de codi Unicode**.
UTF-8, en canvi, és una de les codificacions que permet convertir aquests
caràcters en bytes per emmagatzemar-los o transmetre'ls.

Les funcions `chr()` i `ord()` permeten passar d'un punt de codi al caràcter
corresponent i viceversa:

- `chr(97)` retorna `"a"`
- `ord("a")` retorna `97`

```python
# Exemple

print(chr(76))
print(ord('z'))
```

## Cadenes de text (`str`)

A Python, el text es representa mitjançant objectes de tipus `str`
(*strings* o cadenes de text).

Una cadena és una **seqüència immutable de caràcters Unicode**. Com passa
amb altres seqüències, cada element té una posició o índex i podem utilitzar
operacions com:

- indexació: `cadena[0]`
- *slicing*: `cadena[1:4]`
- longitud: `len(cadena)`
- recorregut amb `for`

La diferència important respecte d'una llista és que una cadena és
**immutable**: un cop creada, no podem modificar directament un dels seus
caràcters.

```python
# Exemple
nom: str = "Eulàlia"

print(nom[0])

for i in range(0, len(nom), 2):
    print(nom[i], end="")

print()
print(nom[0:2])
print(nom[::-1])
```

```python
# però...
nom[3] = "r"
```

## Mètodes i funcions per a les cadenes

Com que les cadenes de caràcters són una de les dades bàsiques amb les que es treballa, coneixerem algunes funcions que ens resultaran molt útils per aquest curs introductori.

### split 

Amb split puc separar una cadena en una llista amb els diferents trossets, indicant el separador.

```python
# Exemple
"456.342.120".split(".")
```

Si no indiquem cap separador, `split()` separa la cadena utilitzant els
**espais en blanc** com a delimitadors.

Això inclou espais, tabuladors o salts de línia, i diversos espais
consecutius es tracten com un únic separador.

```python
# Exemple
"un   dos   tres".split()
```

### join 

El mètode `join()` crea una cadena concatenant diverses cadenes i
utilitzant com a separador la cadena sobre la qual invoquem el mètode.

```python
# Exemple
paraules = ["una", "frase", "curta"]
" ".join(paraules)
```

### strip

El mètode `strip()` retorna una còpia de la cadena eliminant caràcters
del **principi i del final**.

Sense arguments elimina els espais en blanc.

```python
# Exemple
"introducció a Python".strip('nio')
```

```python
" introducció   ".strip()
```

Cal tenir en compte que "nio" no es tracta com una subcadena.
strip() interpreta l'argument com un conjunt de caràcters i elimina
qualsevol combinació d'aquests caràcters mentre apareguin als extrems.

Si volem eliminar exactament un prefix o un sufix, podem utilitzar
removeprefix() o removesuffix().

### `islower()`, `isupper()` i `isalpha()`

Aquests mètodes permeten comprovar característiques dels caràcters
d'una cadena:

- `islower()` retorna `True` si hi ha almenys un caràcter i si tots els caràcters estan en minúscula.
- `isupper()` retorna `True` si hi ha almenys un caràcter i si tots aquests caràcters estan en majúscula.
- `isalpha()` retorna `True` si la cadena no és buida i tots els seus
  caràcters són alfabètics.

`isalpha()` treballa amb caràcters **Unicode**, no només amb les lletres
ASCII. Per exemple, lletres accentuades també poden ser alfabètiques.

### `lower()` i `upper()`

- `lower()` retorna una nova cadena amb els caràcters convertits a minúscules.
- `upper()` retorna una nova cadena amb els caràcters convertits a majúscules.

Com que les cadenes són immutables, aquests mètodes **no modifiquen la
cadena original**, sinó que retornen una cadena nova.

```python
# Exemples

"HOLA".islower()
```

```python
"HOLA".isupper()
```

```python
"HOLA".isalpha()
```

```python
"1HOLA".isalpha()
```

```python
s: str = "HOLA".lower()
print(s)
```

```python
t: str = "Hola".upper()
print(t)
```
